### Transaction Data Findings

- **Normal invoices** are invoices that do not start with `C` or `A`. Rows with positive `Quantity` and positive `Price` represent normal product sales.

- **`C` invoices** indicate cancelled transactions or customer returns. They usually have a negative `Quantity` and positive `Price`, producing a negative transaction value (`Quantity × Price`). These records should be used to calculate return value and net sales. [UCI Online Retail II documentation](https://archive.ics.uci.edu/dataset/502/online%2Bretail%2Bii)

- **`A` invoices** represent administrative adjustments rather than normal customer purchases. They may contain unusual positive or negative values and should be excluded from standard customer-sales KPIs.

- Some normal invoices have a negative `Quantity` with a zero `Price`. These records have a transaction value of zero, so they do not affect sales revenue. They are treated as stock or data adjustments and excluded from sales KPIs.

### KPI Definitions

- **Gross Sales:** Positive-quantity, positive-price normal sales.
- **Returns:** Negative value from `C` invoices.
- **Net Sales:** Gross Sales + Returns.
