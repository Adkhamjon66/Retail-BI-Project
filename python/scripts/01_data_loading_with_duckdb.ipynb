```python
import duckdb
con = duckdb.connect()

con.execute(f"""
    CREATE OR REPLACE VIEW retail  AS
    SELECT *
    FROM read_csv_auto('../../data/clean/online_retail_combined.csv', header = True)
    """
)
```

```python
con.execute("""
    select *
    from retail
    limit 10
""").df()
```
